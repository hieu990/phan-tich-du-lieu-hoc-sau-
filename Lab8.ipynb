3. Viết chương trình Python với thư viện NLTK để kiểm tra danh sách các stopword bằng các 
ngôn ngữ khác nhau

```python
english_stops = stopwords.words('english')
french_stops = stopwords.words('french')

print("Stopwords tiếng Anh (10 từ đầu):", english_stops[:10])
print("Stopwords tiếng Pháp (10 từ đầu):", french_stops[:10])
```

4. Viết chương trình Python với thư viện NLTK để loại bỏ các stopword từ một văn bản đã cho.


```python
text = "This is a simple example to show how to remove stop words from a sentence."
words = nltk.word_tokenize(text)
stop_words_set = set(stopwords.words('english'))

# Lọc bỏ các từ nằm trong danh sách stopword
filtered_sentence = [w for w in words if w.lower() not in stop_words_set]
print("Câu sau khi loại bỏ stopword:", filtered_sentence)
```

5. Viết chương trình Python với thư viện NLTK bỏ qua các stopword từ danh sách các 
stopword

```python
custom_stopwords = set(stopwords.words('english'))

# Giả sử bạn không muốn coi từ 'not' và 'is' là stopword nữa
custom_stopwords.discard('not')
custom_stopwords.discard('is')

filtered_custom = [w for w in words if w.lower() not in custom_stopwords]
print("Câu lọc với danh sách stopword tùy chỉnh:", filtered_custom)
```

6. Viết một chương trình Python với thư viện NLTK để tìm định nghĩa và ví dụ của một từ đã 
cho bằng WordNet từ Wikipedia,

```python
word = "computer"
# Lấy synset (tập hợp các nghĩa) đầu tiên của từ
syn = wordnet.synsets(word)[0]

print(f"Từ: {word}")
print(f"Định nghĩa: {syn.definition()}")
print(f"Ví dụ: {syn.examples()}")
```

7. Viết chương trình Python với thư viện NLTK để tìm tập hợp các từ đồng nghĩa và trái nghĩa 
của một từ nào đó

```python
synonyms = []
antonyms = []

for syn in wordnet.synsets("good"):
    for lemma in syn.lemmas():
        synonyms.append(lemma.name())
        # Nếu có từ trái nghĩa thì thêm vào list
        if lemma.antonyms():
            antonyms.append(lemma.antonyms()[0].name())

print("Đồng nghĩa của 'good':", set(synonyms))
print("Trái nghĩa của 'good':", set(antonyms))
```

8. Viết chương trình Python với thư viện NLTK để có cái nhìn tổng quan về bộ tag, chi tiết của 
một tag cụ thể trong bộ tag và chi tiết về một số bộ tag liên quan, sử dụng biểu thức chính 
quy.

```python
# Xem thông tin về các tag đại diện cho Danh từ (Noun - bắt đầu bằng NN)
print("Chi tiết về các tag Noun:")
nltk.help.upenn_tagset('NN.*')

# Xem thông tin về các tag đại diện cho Động từ (Verb - bắt đầu bằng VB)
print("\nChi tiết về các tag Verb:")
nltk.help.upenn_tagset('VB.*')
```

9. Viết chương trình Python với thư viện NLTK để so sánh sự giống nhau của hai danh từ đã 
cho

```python
# Lấy nghĩa danh từ của 2 từ
cat = wordnet.synsets('cat', pos=wordnet.NOUN)[0]
dog = wordnet.synsets('dog', pos=wordnet.NOUN)[0]

# Tính độ tương đồng Wu-Palmer
similarity_noun = cat.wup_similarity(dog)
print(f"Độ tương đồng giữa Cat và Dog: {similarity_noun}")
```

10. Viết chương trình Python với thư viện NLTK để so sánh sự giống nhau của hai động từ đã 
cho.


```python
# Lấy nghĩa động từ
run = wordnet.synsets('run', pos=wordnet.VERB)[0]
sprint = wordnet.synsets('sprint', pos=wordnet.VERB)[0]

similarity_verb = run.wup_similarity(sprint)
print(f"Độ tương đồng giữa Run và Sprint: {similarity_verb}")
```

11. Viết chương trình Python với thư viện NLTK để tìm số lượng tên nam và nữ trong các tên
kho ngữ liệu. In tên 10 nam và nữ đầu tiên. Lưu ý: Kho văn bản tên chứa tổng cộng khoảng 
2943 nam (male.txt) và 5001 nữ (Female.txt) tên. Kho được biên soạn bởi Kantrowitz, Ross

```python
male_names = names.words('male.txt')
female_names = names.words('female.txt')

print(f"Tổng số tên Nam: {len(male_names)}")
print(f"Tổng số tên Nữ: {len(female_names)}")

print("10 tên Nam đầu tiên:", male_names[:10])
print("10 tên Nữ đầu tiên:", female_names[:10])
```

12. Viết chương trình Python với thư viện NLTK để in 15 kết hợp ngẫu nhiên đầu tiên được gắn 
nhãn nam và được gắn nhãn tên nữ từ kho tên.


```python
# Tạo danh sách các tuple (tên, nhãn giới tính)
labeled_names = ([(name, 'male') for name in male_names] + 
                 [(name, 'female') for name in female_names])

# Trộn ngẫu nhiên
random.shuffle(labeled_names)

print("15 tên ngẫu nhiên kèm nhãn giới tính:")
for name, gender in labeled_names[:15]:
    print(f"{name} - {gender}")
```

13. Viết chương trình Python với thư viện NLTK để trích xuất ký tự cuối cùng của tất cả các tên 
được gắn nhãn và tạo mảng mới với chữ cái cuối cùng của mỗi tên và nhãn được liên kết

```python
# Tạo mảng mới chứa ký tự cuối cùng và nhãn giới tính
features = [(name[-1], gender) for (name, gender) in labeled_names]

print("Mảng đặc trưng (ký tự cuối, giới tính) của 10 mẫu đầu tiên:")
print(features[:10])
```
