1. Xác định và phân loại dữ liệu định tính và định lượng

Dữ liệu định tính (Qualitative): Dùng để phân loại, mô tả đặc tính, không có ý nghĩa tính toán số học.Các biến trong bài: STT (Mã định danh học sinh), GT (Giới tính), DT (Dân tộc), KV (Khu vực), KT (Khối thi).Dữ liệu định lượng (Quantitative): Biểu diễn bằng số thực tế, có ý nghĩa về mặt độ lớn, tính được trung bình, cộng/trừ.Các biến trong bài: Điểm các môn học qua 6 học kỳ (T1 $\rightarrow$ N6) và Điểm thi Đại học (DH1, DH2, DH3).

```python
import pandas as pd

# 1. Sử dụng chính xác hàm pd.read_csv để đọc file CSV
df = pd.read_csv('/dulieuxettuyendaihoc.csv')

# 2. Làm sạch khoảng trắng thừa ở tên các cột
df.columns = df.columns.str.strip()

# 3. Phân loại biến định tính
qualitative_vars = ['STT', 'GT', 'DT', 'KV', 'KT']

# 4. Ép kiểu dữ liệu an toàn từng cột một
for var in qualitative_vars:
    if var in df.columns:
        df[var] = df[var].astype('object')

# 5. Lọc các biến định lượng còn lại
quantitative_vars = [col for col in df.columns if col not in qualitative_vars]

# 6. In kết quả kiểm tra
print("--- ĐÃ KHẮC PHỤC LỖI VÀ PHÂN LOẠI THÀNH CÔNG ---")
print(f"Số biến định tính: {len(qualitative_vars)}")
print(f"Số biến định lượng: {len(quantitative_vars)}")
```

2. Định nghĩa các thang đo phù hợp cho từng biến số

Thang đo định danh (Nominal Scale): Chỉ dùng để phân loại và gắn nhãn, các nhóm bình đẳng, không có thứ tự. Áp dụng: STT, GT, DT, KT.Thang đo thứ bậc (Ordinal Scale): Có tính phân loại và có thứ tự hơn-kém rõ ràng nhưng khoảng cách các bậc không bằng nhau.Áp dụng: KV (Khu vực tuyển sinh - thể hiện mức độ ưu tiên cộng điểm giảm dần: 1 > 2NT > 2 > 3).Thang đo khoảng (Interval Scale): Dữ liệu số có khoảng cách bằng nhau, nhưng không có điểm 0 tuyệt đối (Điểm 0 chỉ là quy ước làm bài thi, không có nghĩa là năng lực học sinh hoàn toàn bằng không).Áp dụng: Toàn bộ các cột điểm số (T1 $\rightarrow$ N6 và DH1 $\rightarrow$ DH3).

```python

thang_do_dict = {}

for var in qualitative_vars:
    if var in df.columns:
        if var == 'KV':
            thang_do_dict[var] = 'Định danh (Nominal) / Thứ bậc (Ordinal)'
        else:
            thang_do_dict[var] = 'Định danh (Nominal)'

for col in quantitative_vars:
    thang_do_dict[col] = 'Khoảng (Interval)'

df_bao_cao_thang_do = pd.DataFrame({
    'Tên Cột/Biến': list(thang_do_dict.keys()),
    'Phân Nhóm Biến': ['Định tính' if c in qualitative_vars else 'Định lượng' for c in thang_do_dict.keys()],
    'Thang Đo Phù Hợp': list(thang_do_dict.values())
})

print("=" * 65)
print("  BẢNG ĐỊNH NGHĨA THANG ĐO CHO TỪNG BIẾN SỐ (BÀI 2)")
print("=" * 65)

display(df_bao_cao_thang_do.iloc[[0, 1, 2, 3, 4, 5, 6, -3, -2, -1]])
```

4. Thống kê dữ liệu thiếu cho cột dân tộc và hiệu chỉnh dữ liệu thiếu như sau: Mặc định thiếu
thì điền giá trị 0.
Hướng dẫn
     1. Lập bảng tần số, tần suất để khảo sát dữ liệu thiếu, bao nhiêu dữ liệu riêng biệt (pandas
unique)
     2. Thực hiện thay thế dữ liệu thiếu bằng phương pháp điền dữ liệu 0


```python
import pandas as pd

df = pd.read_csv('/dulieuxettuyendaihoc.csv')
df.columns = df.columns.str.strip()

print("=" * 70)
print(" 1. KHẢO SÁT DỮ LIỆU TRƯỚC KHI HIỆU CHỈNH")
print("=" * 70)

so_luong_thieu = df['DT'].isna().sum()
print(f"-> Số lượng dòng bị thiếu dữ liệu ở cột 'DT': {so_luong_thieu} dòng.")

cac_gia_tri_rieng_biet = df['DT'].dropna().unique()
print(f"-> Các giá trị riêng biệt hiện có trong cột 'DT': {cac_gia_tri_rieng_biet}")

bang_thong_ke = pd.DataFrame({
    'Tần số': df['DT'].value_counts(dropna=False),
    'Tần suất': df['DT'].value_counts(dropna=False, normalize=True)
})
print("\nBảng tần số và tần suất ban đầu (Có bao gồm dữ liệu thiếu):")
display(bang_thong_ke)


print("\n" + "=" * 70)
print(" 2. THỰC HIỆN HIỆU CHỈNH ĐIỀN GIÁ TRỊ 0")
print("=" * 70)

df['DT'] = df['DT'].fillna(0)

so_luong_thieu_sau = df['DT'].isna().sum()
print(f"-> Số lượng dòng bị thiếu ở cột 'DT' sau khi sửa: {so_luong_thieu_sau} dòng.")

bang_thong_ke_sau = pd.DataFrame({
    'Tần số': df['DT'].value_counts(),
    'Tần suất': df['DT'].value_counts(normalize=True)
})
print("\nBảng tần số và tần suất sau khi hiệu chỉnh (NaN chuyển thành 0.0):")
display(bang_thong_ke_sau)
```

5. Thống kê dữ liệu thiếu cho biến T1 và hiệu chỉnh dữ liệu, lưu ý việc thay thế dữ liệu thiếu sử
dụng phương pháp Mean.
Hướng dẫn
   1. Lập bảng tần số, tần suất để khảo sát dữ liệu thiếu
   2. Thực hiện thay thế dữ liệu thiếu bằng phương pháp Mean

```python
import pandas as pd

df = pd.read_csv('/dulieuxettuyendaihoc.csv')
df.columns = df.columns.str.strip()

print("=" * 70)
print(" 1. KHẢO SÁT DỮ LIỆU THIẾU CỦA BIẾN T1")
print("=" * 70)

so_luong_thieu_t1 = df['T1'].isna().sum()
print(f"-> Số lượng dòng bị thiếu dữ liệu ở cột 'T1': {so_luong_thieu_t1} dòng.")

bang_thong_ke_t1 = pd.DataFrame({
    'Tần số': df['T1'].value_counts(dropna=False),
    'Tần suất': df['T1'].value_counts(dropna=False, normalize=True)
})
print("\nBảng tần số và tần suất của T1 trước khi hiệu chỉnh (Một số dòng đầu):")
display(bang_thong_ke_t1.head(10))


print("\n" + "=" * 70)
print(" 2. THỰC HIỆN THAY THẾ DỮ LIỆU THIẾU BẰNG PHƯƠNG PHÁP MEAN")
print("=" * 70)

gia_tri_mean_t1 = df['T1'].mean()
print(f"-> Giá trị trung bình (Mean) của cột 'T1' là: {gia_tri_mean_t1:.4f}")

df['T1'] = df['T1'].fillna(gia_tri_mean_t1)

so_luong_thieu_t1_sau = df['T1'].isna().sum()
print(f"-> Số lượng dòng bị thiếu dữ liệu ở cột 'T1' sau khi hiệu chỉnh: {so_luong_thieu_t1_sau} dòng.")

print("\n=== ĐÃ HOÀN THÀNH XỬ LÝ BIẾN T1 ===")
```

6. Hãy thực hiện xử lý lần lượt tất cả dữ liệu thiếu cho các biến về điểm số còn lại

```python
import pandas as pd

df = pd.read_csv('/dulieuxettuyendaihoc.csv')
df.columns = df.columns.str.strip()

qualitative_vars = ['STT', 'GT', 'DT', 'KV', 'KT']
quantitative_vars = [col for col in df.columns if col not in qualitative_vars]

print("=" * 80)
print(" TIẾN TRÌNH XỬ LÝ DỮ LIỆU THIẾU CHO CÁC BIẾN ĐIỂM SỐ")
print("=" * 80)

so_cot_da_sua = 0

for col in quantitative_vars:
    so_luong_thieu = df[col].isna().sum()

    if so_luong_thieu > 0:
        mean_value = df[col].mean()

        df[col] = df[col].fillna(mean_value)

        print(f" Cột '{col}': Phát hiện {so_luong_thieu} dòng trống -> Đã điền bằng Mean = {mean_value:.4f}")
        so_cot_da_sua += 1

if so_cot_da_sua == 0:
    print(" Thông báo: Tất cả các cột điểm số hiện tại đều đã đầy đủ 100% dữ liệu.")
    print(" Hệ thống tự động quét qua toàn bộ cấu trúc và sẵn sàng xử lý nếu có dữ liệu trống.")
else:
    print(f"\n Thành công: Đã xử lý triệt để dữ liệu thiếu cho {so_cot_da_sua} cột điểm.")

tong_nan_con_lai = df[quantitative_vars].isna().sum().sum()
print(f" Tổng số dữ liệu thiếu còn lại trên tất cả cột điểm: {tong_nan_con_lai} dòng.")
print("=" * 80)
```

7. Tạo các biến TBM1, TBM2, TBM3 tương ứng với trung bình môn của các năm lớp 10, 11 và 12.
• Công thức tính: TBM = (T*2 + L + H + S + V*2 + X + D + N) / 10

```python
import pandas as pd

df = pd.read_csv('/dulieuxettuyendaihoc.csv')
df.columns = df.columns.str.strip()

tbm_hk1 = (df['T1']*2 + df['L1'] + df['H1'] + df['S1'] + df['V1']*2 + df['X1'] + df['D1'] + df['N1']) / 10
tbm_hk2 = (df['T2']*2 + df['L2'] + df['H2'] + df['S2'] + df['V2']*2 + df['X2'] + df['D2'] + df['N2']) / 10
df['TBM1'] = (tbm_hk1 + tbm_hk2) / 2


tbm_hk3 = (df['T3']*2 + df['L3'] + df['H3'] + df['S3'] + df['V3']*2 + df['X3'] + df['D3'] + df['N3']) / 10
tbm_hk4 = (df['T4']*2 + df['L4'] + df['H4'] + df['S4'] + df['V4']*2 + df['X4'] + df['D4'] + df['N4']) / 10
df['TBM2'] = (tbm_hk3 + tbm_hk4) / 2

tbm_hk5 = (df['T5']*2 + df['L5'] + df['H5'] + df['S5'] + df['V5']*2 + df['X5'] + df['D5'] + df['N5']) / 10
tbm_hk6 = (df['T6']*2 + df['L6'] + df['H6'] + df['S6'] + df['V6']*2 + df['X6'] + df['D6'] + df['N6']) / 10
df['TBM3'] = (tbm_hk5 + tbm_hk6) / 2

print("=" * 75)
print("  ĐÃ TẠO THÀNH CÔNG 3 BIẾN TBM1, TBM2, TBM3 ")
print("=" * 75)

hiển_thị_cột = ['STT', 'TBM1', 'TBM2', 'TBM3']
display(df[hiển_thị_cột].head(5))
```

8. Tạo các biến xếp loại XL1, XL2 và XL3 dựa trên TBM1,TBM2 và TBM3 cho từng năm lớp 10, 11,  12 như sau:
• Nhỏ hơn 5.0 xếp loại: yếu (kí hiệu là Y)
• Từ 5.0 đến dưới 6.5: trung bình (kí hiệu là TB)
• Từ 6.5 đến dưới 8.0: khá (kí hiệu là K)
• Từ 8.0 đến dưới 9.0: giỏi (kí hiệu là G)
• Từ 9.0 trở lên: xuất sắc (kí hiệu là XS)

```python
import pandas as pd

df = pd.read_csv('/dulieuxettuyendaihoc.csv')
df.columns = df.columns.str.strip()

df['TBM1'] = ((df['T1']*2 + df['L1'] + df['H1'] + df['S1'] + df['V1']*2 + df['X1'] + df['D1'] + df['N1'])/10 + (df['T2']*2 + df['L2'] + df['H2'] + df['S2'] + df['V2']*2 + df['X2'] + df['D2'] + df['N2'])/10) / 2
df['TBM2'] = ((df['T3']*2 + df['L3'] + df['H3'] + df['S3'] + df['V3']*2 + df['X3'] + df['D3'] + df['N3'])/10 + (df['T4']*2 + df['L4'] + df['H4'] + df['S4'] + df['V4']*2 + df['X4'] + df['D4'] + df['N4'])/10) / 2
df['TBM3'] = ((df['T5']*2 + df['L5'] + df['H5'] + df['S5'] + df['V5']*2 + df['X5'] + df['D5'] + df['N5'])/10 + (df['T6']*2 + df['L6'] + df['H6'] + df['S6'] + df['V6']*2 + df['X6'] + df['D6'] + df['N6'])/10) / 2

khoang_diem = [0.0, 5.0, 6.5, 8.0, 9.0, 11.0]
nhan_xep_loai = ['Y', 'TB', 'K', 'G', 'XS']

df['XL1'] = pd.cut(df['TBM1'], bins=khoang_diem, labels=nhan_xep_loai, right=False)
df['XL2'] = pd.cut(df['TBM2'], bins=khoang_diem, labels=nhan_xep_loai, right=False)
df['XL3'] = pd.cut(df['TBM3'], bins=khoang_diem, labels=nhan_xep_loai, right=False)

print("=" * 80)
print("  ĐÃ TẠO VÀ XẾP LOẠI HỌC LỰC XL1, XL2, XL3 THÀNH CÔNG ")
print("=" * 80)

# Hiển thị 10 dòng đầu tiên của bảng kết quả so sánh giữa Điểm trung bình và Xếp loại
cot_hien_thi = ['STT', 'TBM1', 'XL1', 'TBM2', 'XL2', 'TBM3', 'XL3']
display(df[cot_hien_thi].head(10))

# Thống kê nhanh số lượng từng loại học lực của năm lớp 10 để kiểm tra tỉ lệ
print("\n Thống kê số lượng học sinh theo từng mức xếp loại của năm lớp 10 (XL1):")
print(df['XL1'].value_counts())
```

9. Tạo các biến US_TBM1, US_TBM2 và US_TBM3 để chuyển điểm trung bình các năm lớp
10, 11 và 12 từ thang điểm 10 của Việt Nam sang thang điểm 4 của Mỹ. Sử dụng phương
pháp Min-Max Normalization

```python
import pandas as pd

df = pd.read_csv('/dulieuxettuyendaihoc.csv')
df.columns = df.columns.str.strip()

df['TBM1'] = ((df['T1']*2 + df['L1'] + df['H1'] + df['S1'] + df['V1']*2 + df['X1'] + df['D1'] + df['N1'])/10 + (df['T2']*2 + df['L2'] + df['H2'] + df['S2'] + df['V2']*2 + df['X2'] + df['D2'] + df['N2'])/10) / 2
df['TBM2'] = ((df['T3']*2 + df['L3'] + df['H3'] + df['S3'] + df['V3']*2 + df['X3'] + df['D3'] + df['N3'])/10 + (df['T4']*2 + df['L4'] + df['H4'] + df['S4'] + df['V4']*2 + df['X4'] + df['D4'] + df['N4'])/10) / 2
df['TBM3'] = ((df['T5']*2 + df['L5'] + df['H5'] + df['S5'] + df['V5']*2 + df['X5'] + df['D5'] + df['N5'])/10 + (df['T6']*2 + df['L6'] + df['H6'] + df['S6'] + df['V6']*2 + df['X6'] + df['D6'] + df['N6'])/10) / 2

df['US_TBM1'] = (df['TBM1'] / 10) * 4
df['US_TBM2'] = (df['TBM2'] / 10) * 4
df['US_TBM3'] = (df['TBM3'] / 10) * 4

print("=" * 85)
print("  ĐÃ CHUYỂN ĐỔI THÀNH CÔNG SANG THANG ĐIỂM 4 CỦA MỸ (US_TBM) ")
print("=" * 85)

cot_hien_thi = ['STT', 'TBM1', 'US_TBM1', 'TBM2', 'US_TBM2', 'TBM3', 'US_TBM3']
display(df[cot_hien_thi].head(5))
```

10. Tạo biến kết quả xét tuyển (kí hiệu là KQXT) nhằm xác định sinh viên đậu (giá trị “1”) và
rớt ( giá trị “0”) vào các khối dựa trên điểm DH1, DH2 và DH3 như sau
• Với khối A, A1 nếu [(DH1*2 + DH2 + DH3)/4] lớn hơn hoặc bằng 5.0 thì đậu,
ngược lại là rớt
• Với khối B nếu [(DH1 + DH2*2 + DH3)/4] lớn hơn hoặc bằng 5.0 thì đậu, ngược
lại là rớt
• Với khối khác nếu [(DH1+ DH2 + DH3)/3] lớn hơn hoặc bằng 5.0 thì đậu, ngược
lại là rớt


```python
import pandas as pd

df = pd.read_csv('/dulieuxettuyendaihoc.csv')
df.columns = df.columns.str.strip()

def tinh_ket_qua_xet_tuyen(row):
    khối_thi = str(row['KT']).strip()
    dh1 = row['DH1']
    dh2 = row['DH2']
    dh3 = row['DH3']

    if khối_thi in ['A', 'A1']:
        diem_xt = (dh1 * 2 + dh2 + dh3) / 4

    elif khối_thi == 'B':
        diem_xt = (dh1 + dh2 * 2 + dh3) / 4

    else:
        diem_xt = (dh1 + dh2 + dh3) / 3

    return 1 if diem_xt >= 5.0 else 0

df['KQXT'] = df.apply(tinh_ket_qua_xet_tuyen, axis=1)
print("=" * 80)
print(" ĐÃ TẠO BIẾN KẾT QUẢ XÉT TUYỂN (KQXT) THÀNH CÔNG ")
print("=" * 80)

cot_hien_thi = ['STT', 'KT', 'DH1', 'DH2', 'DH3', 'KQXT']
display(df[cot_hien_thi].head(10))

print("\n📍 Thống kê số lượng học sinh Đậu (1) và Rớt (0) trên toàn bộ danh sách:")
print(df['KQXT'].value_counts())
```
