```python
# Khai báo biến macbeth trước
macbeth = nltk.corpus.gutenberg.words('shakespeare-macbeth.txt')

# Sau đó mới thực hiện các bước lọc stopword tiếp theo
from nltk.corpus import stopwords
sw = set(stopwords.words('english'))
macbeth_filtered = [w for w in macbeth if w.lower() not in sw]
print("Số lượng từ sau khi lọc stopword:", len(macbeth_filtered))
```

```python
import string
from nltk.corpus import stopwords

# Định nghĩa tập dấu câu
punctuation = set(string.punctuation)

# Lọc cả stopword và dấu câu
macbeth_filtered2 = [w.lower() for w in macbeth if w.lower() not in sw and w.lower() not in punctuation]

# Kiểm tra lại số lượng từ sau khi đã lọc sạch
print("Số lượng từ sau khi lọc stopword và dấu câu:", len(macbeth_filtered2))
```

```python
fd = nltk.FreqDist(macbeth_filtered2)
print("10 từ phổ biến nhất:", fd.most_common(10))
```

```python
# Lấy các từ có độ dài lớn hơn 12 ký tự
long_words = [w for w in macbeth_filtered2 if len(w) > 12]
print("Các từ có độ dài > 12:", sorted(set(long_words)))

# Lấy các từ có chứa chuỗi 'ious'
ious_words = [w for w in macbeth_filtered2 if 'ious' in w]
print("Các từ chứa 'ious':", sorted(set(ious_words)))
```

```python
# Tạo bigrams và liệt kê 15 cặp từ phổ biến nhất
bgrms = nltk.FreqDist(nltk.bigrams(macbeth_filtered2))
print("15 cặp từ (bigrams) phổ biến nhất:", bgrms.most_common(15))
```

```python
nltk.download('wordnet')
from nltk.corpus import wordnet
# Ví dụ tìm từ đồng nghĩa của 'happy'
syns = wordnet.synsets('happy')
print("Từ đồng nghĩa của happy:", syns[0].lemmas()[0].name())
```

9. Viết chương trình Python với thư viện NLTK để so sánh sự giống nhau của hai danh từ đã 
cho. 


```python
import nltk
from nltk.corpus import wordnet

# 1. Tải các gói dữ liệu cần thiết (đảm bảo code chạy không bị lỗi LookupError)
nltk.download('wordnet')
nltk.download('omw-1.4')

# 2. Chọn 2 danh từ cần so sánh (ví dụ: 'cat' và 'dog')
word1 = 'cat'
word2 = 'dog'

# 3. Lấy nghĩa danh từ (Noun) đầu tiên của mỗi từ
# Cú pháp 'từ.n.01' nghĩa là: từ đó + Noun (danh từ) + nghĩa số 01 (phổ biến nhất)
noun1 = wordnet.synset(f'{word1}.n.01')
noun2 = wordnet.synset(f'{word2}.n.01')

# 4. Tính toán độ tương đồng bằng thuật toán Wu-Palmer
similarity = noun1.wup_similarity(noun2)

# 5. In kết quả ra màn hình
print("--- KẾT QUẢ SO SÁNH SỰ GIỐNG NHAU ---")
print(f"Danh từ 1: {word1}")
print(f"Danh từ 2: {word2}")

# Kiểm tra xem có tính được độ tương đồng không
if similarity is not None:
    print(f"Độ tương đồng (Wu-Palmer): {similarity}")
    print(f"Tính theo phần trăm: {similarity * 100:.2f}%")
else:
    print("Không thể so sánh hai từ này.")
```

10. Viết chương trình Python với thư viện NLTK để so sánh sự giống nhau của hai động từ đã cho. 

```python
import nltk
from nltk.corpus import wordnet

# 1. Tải các gói dữ liệu cần thiết (đảm bảo không bị lỗi LookupError)
nltk.download('wordnet')
nltk.download('omw-1.4')

# 2. Chọn 2 động từ cần so sánh (ví dụ: 'run' - chạy và 'sprint' - chạy nước rút)
verb1_word = 'run'
verb2_word = 'sprint'

# 3. Lấy nghĩa động từ (Verb) đầu tiên của mỗi từ
# Cú pháp 'từ.v.01' nghĩa là: từ đó + Verb (động từ) + nghĩa số 01 (phổ biến nhất)
verb1 = wordnet.synset(f'{verb1_word}.v.01')
verb2 = wordnet.synset(f'{verb2_word}.v.01')

# 4. Tính toán độ tương đồng bằng thuật toán Wu-Palmer
similarity = verb1.wup_similarity(verb2)

# 5. In kết quả ra màn hình
print("--- KẾT QUẢ SO SÁNH SỰ GIỐNG NHAU GIỮA 2 ĐỘNG TỪ ---")
print(f"Động từ 1: {verb1_word}")
print(f"Động từ 2: {verb2_word}")

# Kiểm tra xem hệ thống có tính được độ tương đồng không
if similarity is not None:
    print(f"Độ tương đồng (Wu-Palmer): {similarity}")
    print(f"Tính theo phần trăm: {similarity * 100:.2f}%")
else:
    print("Không thể so sánh hai động từ này.")
```

11. Viết chương trình Python với thư viện NLTK để tìm số lượng tên nam và nữ trong các tên kho ngữ liệu. In tên 10 nam và nữ đầu tiên. Lưu ý: Kho văn bản tên chứa tổng cộng khoảng 2943 nam (male.txt) và 5001 nữ (Female.txt) tên. Kho được biên soạn bởi Kantrowitz, Ross.

12. Viết chương trình Python với thư viện NLTK để in 15 kết hợp ngẫu nhiên đầu tiên được gắn 
nhãn nam và được gắn nhãn tên nữ từ kho tên. 

```python
import nltk
import random
from nltk.corpus import names

# 1. Tải gói dữ liệu tên người (nếu chưa tải)
nltk.download('names')

# 2. Tạo danh sách các tên đi kèm với nhãn giới tính
# Sử dụng List Comprehension để gộp tên và nhãn thành một Tuple: (tên, giới_tính)
labeled_names = ([(name, 'male') for name in names.words('male.txt')] + 
                 [(name, 'female') for name in names.words('female.txt')])

# 3. Xáo trộn ngẫu nhiên danh sách vừa tạo
# Lệnh random.shuffle sẽ đảo lộn vị trí các phần tử bên trong danh sách
random.shuffle(labeled_names)

# 4. In ra 15 kết hợp ngẫu nhiên đầu tiên
print("--- 15 KẾT HỢP TÊN NGẪU NHIÊN ĐẦU TIÊN ---")
for i, (name, gender) in enumerate(labeled_names[:15], 1):
    print(f"{i}. Tên: {name} - Nhãn: {gender}")
```

13. Viết chương trình Python với thư viện NLTK để trích xuất ký tự cuối cùng của tất cả các tên 
được gắn nhãn và tạo mảng mới với chữ cái cuối cùng của mỗi tên và nhãn được liên kết. 

```python
import nltk
from nltk.corpus import names

# 1. Tải gói dữ liệu tên người (nếu chưa tải)
nltk.download('names')

# 2. Tạo danh sách các tên đi kèm với nhãn giới tính (giống câu 12)
labeled_names = ([(name, 'male') for name in names.words('male.txt')] + 
                 [(name, 'female') for name in names.words('female.txt')])

# 3. Trích xuất ký tự cuối cùng của mỗi tên và tạo mảng mới
# name[-1] trong Python dùng để lấy ký tự cuối cùng của chuỗi (string)
# Chuyển ký tự cuối thành chữ thường (.lower()) để dữ liệu đồng nhất hơn
featuresets = [(name[-1].lower(), gender) for (name, gender) in labeled_names]

# 4. In ra kiểm tra 10 phần tử đầu tiên và 10 phần tử cuối cùng của mảng mới
print("--- MẢNG ĐẶC TRƯNG (Ký tự cuối, Nhãn) ---")
print(f"10 mẫu đầu tiên (Tên nam): {featuresets[:10]}")
print(f"10 mẫu cuối cùng (Tên nữ): {featuresets[-10:]}")
```
